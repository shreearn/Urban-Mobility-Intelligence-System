```python
# =========================
# SEGMENT 1: FULL ANALYSIS (1000 ROWS)
# =========================

import osmnx as ox
import networkx as nx
import pandas as pd

# =========================
# 1. SAMPLE 1000 ROWS
# =========================
df = eta_df.sample(n=1000, random_state=42).reset_index(drop=True)

# =========================
# 2. CONVERT MILES → KM
# =========================
df['actual_km'] = df['trip_distance'] * 1.60934

# =========================
# 3. CREATE DISTANCE BINS
# =========================
bins = [0, 2, 5, 10, 100]
labels = ['Short (0-2 km)', 'Medium (2-5 km)', 'Long (5-10 km)', 'Very Long (10+ km)']

df['distance_bin'] = pd.cut(df['actual_km'], bins=bins, labels=labels)

# =========================
# 4. LOAD GRAPH
# =========================
try:
    G = ox.load_graphml("nyc.graphml")
except:
    G = ox.graph_from_place("New York City, USA", network_type='drive')
    G = ox.distance.add_edge_lengths(G)
    ox.save_graphml(G, "nyc.graphml")

# =========================
# 5. SHORTEST DISTANCE FUNCTION
# =========================
def get_shortest_distance(row):
    try:
        orig = (row['pickup_latitude'], row['pickup_longitude'])
        dest = (row['dropoff_latitude'], row['dropoff_longitude'])

        orig_node = ox.distance.nearest_nodes(G, X=orig[1], Y=orig[0])
        dest_node = ox.distance.nearest_nodes(G, X=dest[1], Y=dest[0])

        return nx.shortest_path_length(G, orig_node, dest_node, weight='length') / 1000

    except:
        return None

# =========================
# 6. COMPUTE FOR ALL (⚠️ HEAVY)
# =========================
df['shortest_km'] = df.apply(get_shortest_distance, axis=1)

# drop failed
df = df.dropna(subset=['shortest_km'])

# =========================
# 7. COMPARISON
# =========================
df['difference'] = df['shortest_km'] - df['actual_km']
df['shortest_gt_actual'] = df['shortest_km'] > df['actual_km']

# =========================
# 8. BIN-WISE ANALYSIS
# =========================
summary = df.groupby('distance_bin').agg(
    total_trips=('shortest_gt_actual', 'count'),
    cases_shortest_gt_actual=('shortest_gt_actual', 'sum')
)

summary['percentage_shortest_gt_actual'] = (
    summary['cases_shortest_gt_actual'] / summary['total_trips']
) * 100

print("\n========= FINAL BIN-WISE ANALYSIS (1K) =========\n")
print(summary)
```

```python
# WHY DISCREPANCIES? (EXPLORATORY INSIGHTS)

# 1. GPS / pickup-drop inaccuracies : Coordinates may not lie exactly on roads, Snapping to nearest node/intersection introduces error
# 2. Taxi route ≠ shortest path : take detours, avoid traffic ,follow passenger instructions
# 3. Meter / data inaccuracies : Taxi meter rounding, sensor errors, aggregation issues
# 4. Map mismatch (OSM vs real world) : Missing roads, wrong restrictions, outdated data
```

```python
# =========================
# SEGMENT 2: RANDOM SAMPLING FROM EACH BIN
# =========================

print("\n========= RANDOM SAMPLES FROM EACH BIN =========\n")

final_samples = []

for bin_name in df['distance_bin'].dropna().unique():

    bin_data = df[df['distance_bin'] == bin_name]

    # take up to 5 samples
    sample_n = min(5, len(bin_data))

    sampled = bin_data.sample(n=sample_n, random_state=42)

    print(f"\n--- {bin_name} ---")
    print(f"Selected {sample_n} samples\n")

    print(sampled[['actual_km', 'shortest_km', 'difference']])

    final_samples.append(sampled)

# combine
final_df = pd.concat(final_samples).reset_index(drop=True)

print("\n========= FINAL SELECTED TRIPS =========\n")
print(final_df[['actual_km', 'shortest_km', 'difference', 'distance_bin']])
```

```python
import json
import numpy as np
import os
import pandas as pd
import webbrowser

# --- JSON ENCODER ---
class NpEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, np.integer): return int(obj)
        if isinstance(obj, np.floating): return float(obj)
        if isinstance(obj, np.ndarray): return obj.tolist()
        return super().default(obj)

# --- DATA PREP ---
unique_bins = sorted(final_df['distance_bin'].dropna().unique())
palette = ["#22c55e", "#3b82f6", "#f59e0b", "#ef4444"]

bin_colors = {b: palette[i % len(palette)] for i, b in enumerate(unique_bins)}

all_trips_data = {}

for bin_name in unique_bins:
    subset = final_df[final_df['distance_bin'] == bin_name]
    trips = []

    for i in range(min(10, len(subset))):
        try:
            row = subset.iloc[i]

            orig = ox.distance.nearest_nodes(G, X=row['pickup_longitude'], Y=row['pickup_latitude'])
            dest = ox.distance.nearest_nodes(G, X=row['dropoff_longitude'], Y=row['dropoff_latitude'])

            route = ox.shortest_path(G, orig, dest, weight='length')

            if route is None or len(route) < 2:
                continue

            coords = [[float(G.nodes[n]['y']), float(G.nodes[n]['x'])] for n in route]

            pickup_dt = pd.to_datetime(row['tpep_pickup_datetime'])

            trips.append({
                "coords": coords,
                "metrics": {
                    "lgbm": round(row['lgbm_v3_pred_min'], 2),
                    "actual": round(row['actual_duration_min'], 2),
                    "dist_osm": round(row['shortest_km'], 2),
                    "dist_act": round(row['actual_km'], 2),
                },
                "date": pickup_dt.strftime('%b %d, %Y'),
                "time": pickup_dt.strftime('%I:%M %p'),
                "color": bin_colors[bin_name]
            })

        except:
            continue

    if len(trips) > 0:
        all_trips_data[str(bin_name)] = trips

json_payload = json.dumps(all_trips_data, cls=NpEncoder)

# --- HTML ---
html_template = """
<!DOCTYPE html>
<html>
<head>
<meta charset="utf-8"/>
<title>Route Dashboard</title>

<link rel="stylesheet" href="https://unpkg.com/leaflet@1.7.1/dist/leaflet.css"/>
<script src="https://unpkg.com/leaflet@1.7.1/dist/leaflet.js"></script>
<script src="https://cdn.jsdelivr.net/npm/leaflet-ant-path@1.3.0/dist/leaflet-ant-path.js"></script>

<style>
body { margin:0; background:#020617; }
#map { width:100%; height:100vh; }

.panel {
    position:absolute;
    z-index:1000;
    background:rgba(0,0,0,0.7);
    padding:12px;
    border-radius:8px;
    color:white;
}

#controls { top:20px; left:20px; width:250px; }
#hud { bottom:20px; left:20px; width:250px; }

select, input {
    width:100%;
    margin-top:5px;
    margin-bottom:10px;
}
</style>
</head>

<body>

<div id="map"></div>

<div id="controls" class="panel">
<b>Controls</b><br>

<select id="zone"></select>

<div>Sample: <span id="sampleNum">1</span></div>

<input type="range" id="slider" min="0" max="0" value="0">
</div>

<div id="hud" class="panel"></div>

<script>

const data = __DATA__;

const map = L.map('map').setView([40.7128, -74.0060], 12);

L.tileLayer('https://{s}.basemaps.cartocdn.com/dark_all/{z}/{x}/{y}{r}.png').addTo(map);

const zone = document.getElementById("zone");
const slider = document.getElementById("slider");

let pathLayer, startMarker, endMarker;
let glowLayerLocal = null;
let mainLayerLocal = null;

let animationVersion = 0;

const zones = Object.keys(data);

// dropdown
zones.forEach(z => {
    let opt = document.createElement("option");
    opt.value = z;
    opt.text = z + " (" + data[z].length + ")";
    zone.appendChild(opt);
});

function updateSlider() {
    const z = zone.value;
    slider.max = data[z].length - 1;
    slider.value = 0;
    document.getElementById("sampleNum").innerHTML = 1;
}

function updateMap() {

    animationVersion++; //  kill previous animations
    const currentVersion = animationVersion;

    const z = zone.value;
    const i = parseInt(slider.value);
    const trip = data[z][i];
    if (!trip) return;

    document.getElementById("sampleNum").innerHTML = i + 1;

    // remove old
    if(pathLayer) map.removeLayer(pathLayer);
    if(startMarker) map.removeLayer(startMarker);
    if(endMarker) map.removeLayer(endMarker);
    if(glowLayerLocal) map.removeLayer(glowLayerLocal);
    if(mainLayerLocal) map.removeLayer(mainLayerLocal);

    let drawn = [];

    glowLayerLocal = L.polyline([], {
        color: trip.color,
        weight: 16,
        opacity: 0.3
    }).addTo(map);

    mainLayerLocal = L.polyline([], {
        color: trip.color,
        weight: 6
    }).addTo(map);

    let step = 0;

    function drawStep() {

        if (currentVersion !== animationVersion) return;

        if (step >= trip.coords.length) {

            if (currentVersion !== animationVersion) return;

            pathLayer = L.polyline.antPath(trip.coords, {
                color: trip.color,
                weight: 6,
                delay: 1000
            }).addTo(map);

            return;
        }

        drawn.push(trip.coords[step]);

        glowLayerLocal.setLatLngs(drawn);
        mainLayerLocal.setLatLngs(drawn);

        step++;

        setTimeout(drawStep, 35);
    }

    drawStep();

    // markers with S/E
    startMarker = L.marker(trip.coords[0], {
        icon: L.divIcon({
            className: '',
            html: `<div style="
                width:28px;height:28px;border:3px solid #22c55e;border-radius:50%;
                display:flex;align-items:center;justify-content:center;
                color:#22c55e;font-weight:bold;font-size:12px;">S</div>`
        })
    }).addTo(map);

    endMarker = L.marker(trip.coords[trip.coords.length-1], {
        icon: L.divIcon({
            className: '',
            html: `<div style="
                width:28px;height:28px;border:3px solid #ef4444;border-radius:50%;
                display:flex;align-items:center;justify-content:center;
                color:#ef4444;font-weight:bold;font-size:12px;">E</div>`
        })
    }).addTo(map);

    const err = (trip.metrics.lgbm - trip.metrics.actual).toFixed(2);

    document.getElementById("hud").innerHTML = 
        "Date: " + trip.date + "<br>" +
        "Time: " + trip.time + "<br><br>" +
        "ETA: " + trip.metrics.lgbm + " min<br>" +
        "Actual: " + trip.metrics.actual + " min<br>" +
        "Error: " + err + " min<br><br>" +
        "OSM: " + trip.metrics.dist_osm + " km<br>" +
        "Actual Dist: " + trip.metrics.dist_act + " km";

    map.fitBounds(trip.coords);
}

// events
zone.addEventListener("change", () => {
    updateSlider();
    updateMap();
});

slider.addEventListener("input", updateMap);

// init
zone.value = zones[0];
updateSlider();
updateMap();

</script>

</body>
</html>
"""

html_template = html_template.replace("__DATA__", json_payload)

path = os.path.join(os.getcwd(), "dashboard.html")

with open(path, "w", encoding="utf-8") as f:
    f.write(html_template)

print("Saved at:", path)
webbrowser.open("file://" + path)
```
